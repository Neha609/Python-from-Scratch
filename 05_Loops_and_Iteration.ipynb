# 1. Why Iteration Is Needed

Imagine processing five candidate scores manually:

    process(candidate_1)
    process(candidate_2)
    process(candidate_3)
    process(candidate_4)
    process(candidate_5)

This approach becomes impossible when the number of records grows to hundreds, thousands, or millions.

Loops allow us to express the idea:

> Perform the same operation for every relevant item.

Python provides two primary loop structures:

- `for`
- `while`

# 2. The `for` Loop

A `for` loop processes items from an iterable one at a time.

### General structure

    for item in iterable:
        statements

During each iteration, the next value is assigned to the loop variable.

```python
candidate_scores = [78, 92, 66, 85]

for score in candidate_scores:
    print(score)
```

## Processing Values During Iteration

A loop becomes useful when an operation is performed for every value.

Here we classify candidate scores using previously learned conditional logic without reteaching conditionals themselves.

```python
candidate_scores = [78, 92, 66, 85]

for score in candidate_scores:
    if score >= 75:
        print(f"{score}: Qualified")
    else:
        print(f"{score}: Not Qualified")
```

# 3. `range()`

`range()` generates a sequence of integers commonly used for controlled repetition.

Common forms include:

- `range(stop)`
- `range(start, stop)`
- `range(start, stop, step)`

The stop value is excluded.

```python
for number in range(5):
    print(number)
```

```python
for number in range(2, 10, 2):
    print(number)
```

# 4. Iterating Over Sequences

Python's `for` loop works directly with iterable objects.

For this notebook we focus only on **iteration behavior** rather than the internal details of individual data structures.

Those structures are covered separately later.

```python
employee_names = ["Aarav", "Neha", "Riya"]

for employee in employee_names:
    print(f"Processing employee: {employee}")
```

```python
text = "AI"

for character in text:
    print(character)
```

# 5. Counters

A counter tracks how many times a particular event occurs.

Typical applications include:

- counting qualified candidates,
- counting successful API calls,
- counting missing records,
- counting model predictions above a threshold.

```python
candidate_scores = [78, 92, 66, 85, 59]

qualified_count = 0

for score in candidate_scores:
    if score >= 75:
        qualified_count += 1

print(f"Qualified Candidates: {qualified_count}")
```

# 6. Accumulators

An accumulator gradually combines values during iteration.

Common examples include:

- totals,
- sums,
- cumulative costs,
- aggregate scores.

The accumulator usually starts with an initial value and is updated during every relevant iteration.

```python
candidate_scores = [78, 92, 66, 85]

total_score = 0

for score in candidate_scores:
    total_score += score

average_score = total_score / len(candidate_scores)

print(f"Total Score: {total_score}")
print(f"Average Score: {average_score:.2f}")
```

# 7. `while` Loops

A `while` loop repeats as long as its condition remains `True`.

### General structure

    while condition:
        statements

`while` is especially useful when the number of repetitions is not known in advance.

```python
attempt = 1

while attempt <= 3:
    print(f"Processing attempt {attempt}")
    attempt += 1
```

## `for` vs `while`

Use a `for` loop when you are primarily iterating through an existing sequence or a known range.

Use a `while` loop when repetition depends on a condition that changes during execution.

Examples:

- process every employee → `for`
- retry until successful → `while`

# 8. Avoiding Infinite Loops

A `while` loop becomes infinite when its condition never becomes `False`.

Always verify that something inside the loop eventually changes the controlling condition.

```python
remaining_attempts = 3

while remaining_attempts > 0:
    print(f"Attempts remaining: {remaining_attempts}")
    remaining_attempts -= 1

print("No attempts remaining.")
```

# 9. `break`

`break` immediately terminates the nearest enclosing loop.

It is useful when continuing iteration is unnecessary after a target condition has been reached.

Examples include:

- finding the first matching candidate,
- stopping after an API succeeds,
- terminating a search when a result is found.

```python
candidate_scores = [61, 68, 73, 91, 84]

for score in candidate_scores:
    if score >= 90:
        print(f"High-scoring candidate found: {score}")
        break
```

# 10. `continue`

`continue` skips the remaining statements in the current iteration and moves directly to the next iteration.

It is useful when certain records should be ignored without terminating the entire loop.

```python
scores = [82, None, 91, None, 76]

for score in scores:
    if score is None:
        continue

    print(f"Processing score: {score}")
```

# 11. `pass`

`pass` performs no action.

It is mainly used as a placeholder when Python requires a statement syntactically but implementation will be added later.

`pass` does **not** skip the rest of the iteration like `continue`.

```python
records = ["candidate", "employee", "model"]

for record in records:
    if record == "employee":
        pass

    print(f"Processed: {record}")
```

# 12. Loop `else`

Python loops can contain an `else` block.

The `else` block runs when the loop completes normally.

If the loop terminates through `break`, the `else` block does not execute.

This is particularly useful for search-style logic.

```python
candidate_scores = [65, 72, 81, 88]

for score in candidate_scores:
    if score >= 90:
        print("Exceptional candidate found.")
        break
else:
    print("No candidate scored 90 or above.")
```

# 13. `enumerate()`

Sometimes we need both:

- the item,
- its position.

`enumerate()` provides both values directly and is generally preferable to manually maintaining an index counter.

```python
candidates = ["Aarav", "Neha", "Riya"]

for position, candidate in enumerate(candidates, start=1):
    print(f"{position}. {candidate}")
```

# 15. Nested Loops

A loop can exist inside another loop.

The inner loop completes its iterations for each iteration of the outer loop.

Nested loops are useful for combinations, matrix-style processing, and hierarchical data.

However, they can become computationally expensive when working with large datasets.

```python
candidates = ["Aarav", "Neha"]
skills = ["Python", "SQL", "ML"]

for candidate in candidates:
    print(f"\nSkills being evaluated for {candidate}:")

    for skill in skills:
        print(f"- {skill}")
```

# 16. Sentinel-Controlled Loops

A sentinel is a special value used to signal that repetition should stop.

Sentinel-controlled loops are useful when the number of inputs is unknown.

Examples include:

- command-line applications,
- user input processing,
- message queues,
- streaming-style workflows.

```python
commands = ["process", "process", "process", "stop"]

index = 0
command = commands[index]

while command != "stop":
    print(f"Executing command: {command}")

    index += 1
    command = commands[index]

print("Processing stopped.")
```

# 17. Practical Example — Model Predictions

Suppose a machine-learning model returns several prediction confidence scores.

We want to:

- process each prediction,
- count high-confidence predictions,
- identify predictions requiring review.

```python
prediction_confidences = [0.95, 0.72, 0.88, 0.61, 0.93]

high_confidence_count = 0

for confidence in prediction_confidences:
    if confidence >= 0.90:
        status = "High Confidence"
        high_confidence_count += 1
    elif confidence >= 0.70:
        status = "Review"
    else:
        status = "Low Confidence"

    print(f"{confidence:.2f} -> {status}")

print(f"\nHigh-confidence predictions: {high_confidence_count}")
```

# 18. Practical Example — API Results

Applications frequently process multiple API responses.

A loop can inspect each result and apply response-specific handling.

```python
api_status_codes = [200, 200, 404, 500, 200]

successful_requests = 0

for status_code in api_status_codes:
    if 200 <= status_code < 300:
        successful_requests += 1
        print(f"{status_code}: Success")
    else:
        print(f"{status_code}: Requires attention")

print(f"\nSuccessful Requests: {successful_requests}")
```

# 19. Practical Example — Processing Text Records

Iteration is fundamental to text-processing pipelines.

A system may need to inspect multiple pieces of text before sending them to NLP, embedding, retrieval, or Generative AI pipelines.

```python
text_records = [
    "Python for Machine Learning",
    "",
    "Retrieval-Augmented Generation",
    "Agentic AI"
]

for record in text_records:
    if not record:
        continue

    print(f"Processing: {record}")
```

# 20. Common Loop Mistakes

### Modifying the control variable incorrectly

A `while` loop may never terminate if its controlling state is not updated.

### Off-by-one errors

Remember that the stop value of `range()` is excluded.

### Using `break` when `continue` is intended

- `break` terminates the loop.
- `continue` skips only the current iteration.

### Unnecessary index-based iteration

Avoid manually indexing a collection when direct iteration is sufficient.

### Excessive nesting

Nested loops can reduce readability and may increase execution cost significantly.

### Repeating work unnecessarily

Move calculations outside a loop when their result does not change between iterations.

# 21. Pythonic Iteration

Python usually favors direct iteration over manual index management.

Instead of thinking:

> For every integer index, fetch the corresponding item.

Prefer thinking:

> For every item in the collection, process the item directly.

```python
employees = ["Aarav", "Neha", "Riya"]

# Less Pythonic
for index in range(len(employees)):
    print(employees[index])

print("---")

# Preferred
for employee in employees:
    print(employee)
```

## When an Index Is Actually Needed

If both the position and value are required, use `enumerate()` instead of manually generating indexes.

```python
employees = ["Aarav", "Neha", "Riya"]

for index, employee in enumerate(employees, start=1):
    print(f"Employee {index}: {employee}")
```

# 22. Loop Efficiency Basics

Iteration performance becomes important when processing large datasets.

Some basic principles are:

- avoid unnecessary nested loops,
- avoid repeating expensive calculations,
- stop searching with `break` when the required result is already found,
- use built-in Python operations where appropriate,
- avoid processing records that can be skipped early.

Later notebooks will introduce more advanced approaches such as comprehensions, generators, NumPy vectorization, and optimized data-processing patterns.

```python
candidate_scores = [52, 63, 74, 98, 81, 92]

for score in candidate_scores:
    if score >= 95:
        print(f"Exceptional score found: {score}")
        break
```

# 25. Data Science, ML and AI Relevance

Process records, features, files, batches, or validation rules.

### NLP
