# 1. Operators and Expressions

An **operator** is a symbol or keyword that performs an operation on one or more values.

Examples:

- `+` → addition
- `>` → comparison
- `and` → logical evaluation
- `in` → membership check

The values on which an operator acts are called **operands**.

Example:

`10 + 5`

Here:

- `10` and `5` are operands.
- `+` is the operator.

An **expression** is a combination of values, variables, operators, or function calls that Python evaluates to produce a result.

Examples:

`10 + 5`

`salary > 50000`

`score >= 80 and experience >= 3`

```python
# Basic expressions

print(10 + 5)
print(20 > 10)
print(8 * 4 + 2)

salary = 60000
experience = 4

eligible = salary > 50000 and experience >= 3

print("Eligible:", eligible)
```

# 2. Arithmetic Operators

Arithmetic operators perform mathematical calculations.

| Operator | Meaning |
|---|---|
| `+` | Addition |
| `-` | Subtraction |
| `*` | Multiplication |
| `/` | True division |
| `//` | Floor division |
| `%` | Modulus / remainder |
| `**` | Exponentiation |

Arithmetic operators are commonly used in:

- statistical calculations,
- feature engineering,
- financial calculations,
- ML metrics,
- data transformations,
- scoring systems.

```python
a = 17
b = 5

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("True division:", a / b)
print("Floor division:", a // b)
print("Remainder:", a % b)
print("Exponentiation:", a ** b)
```

## `/` vs `//`

Python provides two division operators.

### `/` — True Division

Returns the mathematical division result.

Example:

`10 / 4 → 2.5`

### `//` — Floor Division

Returns the mathematical floor of the division result.

Example:

`10 // 4 → 2`

An important detail is that floor division does **not simply remove the decimal part**.

For negative numbers:

`-10 // 4 → -3`

because Python rounds downward toward negative infinity.

```python
print("10 / 4  =", 10 / 4)
print("10 // 4 =", 10 // 4)

print("-10 / 4  =", -10 / 4)
print("-10 // 4 =", -10 // 4)
```

## Modulus Operator `%`

The modulus operator returns the **remainder after division**.

Example:

`17 % 5 → 2`

Common applications include:

- checking even or odd numbers,
- cyclic calculations,
- bucket assignment,
- periodic processing,
- rotating indexes.

```python
number = 18

print("18 % 5:", number % 5)
print("Even:", number % 2 == 0)
print("Odd:", number % 2 != 0)

employee_id = 27
bucket = employee_id % 4

print("Assigned bucket:", bucket)
```

## Exponentiation `**`

The `**` operator raises a value to a power.

Examples:

`2 ** 3 → 8`

`5 ** 2 → 25`

Exponentiation frequently appears in:

- variance calculations,
- distance calculations,
- loss functions,
- mathematical models,
- polynomial features.

```python
value = 5

print("Square:", value ** 2)
print("Cube:", value ** 3)

x1 = 3
x2 = 7

squared_difference = (x1 - x2) ** 2

print("Squared difference:", squared_difference)
```

# 4. Equality `==` vs Identity `is`

This is an important Python distinction.

## `==` — Equality

Checks whether two objects contain **equal values**.

## `is` — Identity

Checks whether two references point to the **same object**.

Therefore:

`a == b`

asks:

> Do these objects have equal values?

Whereas:

`a is b`

asks:

> Are these exactly the same object?

For normal value comparison, use `==`.

A very common and recommended identity check is:

`value is None`

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print("a == b:", a == b)
print("a is b:", a is b)

print("a == c:", a == c)
print("a is c:", a is c)
```

```python
result = None

print(result is None)
print(result is not None)
```

# 5. Logical Operators

Python provides three logical operators:

- `and`
- `or`
- `not`

They are commonly used to combine or reverse conditions.

Examples:

`age >= 18 and verified`

`is_admin or is_manager`

`not blocked`

One important Python behavior is:

> `and` and `or` do not necessarily return `True` or `False`.

They can return one of their original operands.

```python
age = 25
verified = True

print(age >= 18 and verified)

is_admin = False
is_manager = True

print(is_admin or is_manager)

blocked = False

print(not blocked)
```

## How `and` Works

For:

`A and B`

Python evaluates `A` first.

- If `A` is falsy, Python returns `A`.
- If `A` is truthy, Python returns `B`.

Examples:

`0 and 100 → 0`

`"Python" and "AI" → "AI"`

This explains why `and` does not always return a Boolean value.

```python
print(0 and 100)
print(10 and 100)

print("" and "Python")
print("Python" and "AI")
```

## How `or` Works

For:

`A or B`

Python evaluates `A` first.

- If `A` is truthy, Python returns `A`.
- If `A` is falsy, Python returns `B`.

Examples:

`None or "default" → "default"`

`"admin" or "guest" → "admin"`

This behavior is frequently used for fallback values.

```python
print(None or "default")
print("" or "unknown")
print("admin" or "guest")
print(0 or 100)
```

## Pythonic Default-Value Pattern

A common Python pattern is:

`value = supplied_value or default_value`

Example:

`display_name = user_name or "Guest"`

However, remember that `or` treats **all falsy values** as absent.

That includes:

- `0`
- `False`
- `""`
- empty containers
- `None`

Therefore, this pattern should only be used when replacing all falsy values is intentional.

```python
user_name = ""

display_name = user_name or "Guest"

print(display_name)

# Be careful when 0 is a valid value
threshold = 0
effective_threshold = threshold or 10

print(effective_threshold)
```

# 6. Short-Circuit Evaluation

Python evaluates logical expressions from left to right.

With `and`:

If the first operand is falsy, Python already knows the entire expression cannot be truthy.

Therefore, the remaining expression may not be evaluated.

With `or`:

If the first operand is truthy, Python already has a successful value.

Therefore, the remaining expression may not be evaluated.

This behavior is called **short-circuit evaluation**.

```python
denominator = 0

safe_check = denominator != 0 and 100 / denominator > 5

print(safe_check)
```

Short-circuiting is useful for:

- preventing runtime errors,
- avoiding unnecessary calculations,
- checking prerequisites before operations,
- preventing unnecessary API calls,
- avoiding unnecessary model inference,
- implementing cache-first logic.

Example:

`cached_result or expensive_operation()`

If `cached_result` is already available and truthy, the expensive operation does not need to execute.

```python
cached_result = "Existing response"

result = cached_result or print("Expensive operation executed")

print("Result:", result)
```

# 7. Assignment and Compound Assignment

The standard assignment operator is:

`=`

Example:

`total = price * quantity`

Python also provides compound assignment operators.

| Operator | Equivalent |
|---|---|
| `+=` | `x = x + value` |
| `-=` | `x = x - value` |
| `*=` | `x = x * value` |
| `/=` | `x = x / value` |
| `//=` | `x = x // value` |

# 10. Bitwise Operators

Bitwise operators operate on the binary representation of integers.

| Operator | Meaning |
|---|---|
| `&` | Bitwise AND |
| `|` | Bitwise OR |
| `^` | Bitwise XOR |
| `~` | Bitwise NOT |
| `<<` | Left shift |
| `>>` | Right shift |

Bitwise operations are less common in everyday Python code but are useful in:

- flags and permissions,
- binary protocols,
- image processing,
- systems programming,
- compact state representation.

```python
a = 6   # Binary: 110
b = 3   # Binary: 011

print("AND:", a & b)
print("OR:", a | b)
print("XOR:", a ^ b)
print("NOT:", ~a)
print("Left Shift:", a << 1)
print("Right Shift:", a >> 1)
```

`011`

`010`

`6 & 3 → 2`

# 11. Operator Precedence

When multiple operators appear in one expression, Python follows precedence rules.

A simplified high-to-low order is:

1. Parentheses `()`
2. Exponentiation `**`
3. Unary operators
4. `*`, `/`, `//`, `%`
5. `+`, `-`
6. Comparison operators
7. `not`
8. `and`
9. `or`

Instead of depending heavily on memorized precedence rules, use parentheses when they improve readability.

```python
result = 10 + 5 * 2

print(result)

result_with_parentheses = (10 + 5) * 2

print(result_with_parentheses)
```

```python
result = 2 ** 3 * 4

print(result)

comparison = 10 + 5 > 12 and 3 * 2 == 6

print(comparison)
```

## Precedence-Related Bugs

Consider:

`score > 80 and experience >= 3 or certified`

Python evaluates `and` before `or`.

Even when the result is technically correct, the expression can be difficult to read.

A clearer version is:

`(score > 80 and experience >= 3) or certified`

### Pythonic Practice

Use parentheses whenever they make the intended business logic easier to understand.

Readable code is preferable to unnecessarily clever code.

```python
score = 75
experience = 4
certified = True

without_parentheses = score > 80 and experience >= 3 or certified

with_parentheses = (score > 80 and experience >= 3) or certified

print(without_parentheses)
print(with_parentheses)
```

# 12. Associativity

When operators have the same precedence, **associativity** determines how the expression is grouped.

Most arithmetic operators are evaluated from left to right.

Example:

`20 / 5 * 2`

behaves like:

`(20 / 5) * 2`

Exponentiation is an important exception.

`**` is right-associative.

Therefore:

`2 ** 3 ** 2`

behaves like:

`2 ** (3 ** 2)`

```python
print(20 / 5 * 2)

print(2 ** 3 ** 2)

print((2 ** 3) ** 2)
print(2 ** (3 ** 2))
```

# 13. Chained Comparisons

Python supports chained comparisons.

Instead of:

`age >= 18 and age <= 60`

you can write:

`18 <= age <= 60`

This is concise and Pythonic.

Chained comparisons are useful for:

- range validation,
- probability ranges,
- confidence scores,
- thresholds,
- business-rule validation.

```python
age = 32

print(18 <= age <= 60)

confidence = 0.87

print(0.0 <= confidence <= 1.0)
```

## How Chained Comparisons Work

Consider:

`a < b < c`

Conceptually, this means:

`a < b and b < c`

However, Python handles chained comparisons directly and evaluates the middle expression only once.

This makes chained comparisons both readable and efficient.

# 14. Expression Evaluation

Expressions can combine several operators.

Example:

`(revenue - cost) / revenue * 100`

Python evaluates the expression according to:

1. parentheses,
2. operator precedence,
3. associativity.

The resulting value can then be:

- assigned to a variable,
- compared against another value,
- passed to another operation,
- or used in larger calculations.

```python
revenue = 250000
cost = 175000

profit_margin = (revenue - cost) / revenue * 100

print("Profit Margin:", profit_margin)
```

# 15. Practical Business Example — Candidate Eligibility

Suppose a recruitment process uses the following rules:

- technical score must be at least `75`,
- experience must be at least `3` years,
- candidate must not be blocked,
- candidate location must be supported.

This example combines:

- comparison operators,
- logical operators,
- membership operators,
- expression evaluation.

```python
technical_score = 82
experience_years = 5
blocked = False
location = "Hyderabad"

supported_locations = ["Hyderabad", "Bengaluru", "Pune"]

eligible = (
    technical_score >= 75
    and experience_years >= 3
    and not blocked
    and location in supported_locations
)

print("Candidate Eligible:", eligible)
```

# 16. Practical Data Example — Data Quality Validation

Consider an employee record received from an enterprise system.

A simplified validation rule may require:

- employee ID to exist,
- salary to be non-negative,
- status to belong to the allowed categories,
- confidence score to remain between `0` and `1`.

Operators make such validation rules concise and readable.

```python
employee_id = "E1024"
salary = 85000
status = "Active"
confidence_score = 0.94

valid_statuses = ["Active", "Inactive", "Leave"]

is_valid = (
    employee_id is not None
    and salary >= 0
    and status in valid_statuses
    and 0 <= confidence_score <= 1
)

print("Record Valid:", is_valid)
```

# 19. Common Operator Mistakes

## Mistake 1 — Using `is` Instead of `==`

Use:

`value == 1000`

for value comparison.

Use `is` mainly for identity checks such as:

`value is None`

---

## Mistake 2 — Confusing `/` and `//`

`7 / 2 → 3.5`

`7 // 2 → 3`

---

## Mistake 3 — Assuming `and` and `or` Always Return Boolean Values

`"Python" and "AI"`

returns:

`"AI"`

---

## Mistake 4 — Ignoring Operator Precedence

Prefer:

`(a > 10 and b < 5) or c`

when parentheses clarify intent.

---

## Mistake 5 — Writing Long Range Conditions

Instead of:

`score >= 0 and score <= 1`

prefer:

`0 <= score <= 1`

---

## Mistake 6 — Forgetting Negative Floor Division

`-7 // 2 → -4`

not `-3`.

---

## Mistake 7 — Using `or` for Defaults When Zero Is Valid

`0 or 100`

returns `100`.

Therefore, the `or` fallback pattern may be inappropriate when `0` is a meaningful value.

```python
print("7 / 2 =", 7 / 2)
print("7 // 2 =", 7 // 2)

print("-7 // 2 =", -7 // 2)

print('"Python" and "AI" =', "Python" and "AI")

valid_value = 0
fallback = 100

print("0 or 100 =", valid_value or fallback)
```

# 22. Mini Practice Exercise

An AI system should trigger human review when:

- model confidence is below `0.70`, **or**
- risk score is at least `80`,

and the request has **not already been reviewed**.

Create an expression that determines whether human review is required.

Pay attention to logical grouping and operator precedence.

```python
model_confidence = 0.66
risk_score = 72
already_reviewed = False

requires_review = (
    (model_confidence < 0.70 or risk_score >= 80)
    and not already_reviewed
)

print("Human Review Required:", requires_review)
```
