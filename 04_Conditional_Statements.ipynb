# 1. Boolean Conditions

A conditional statement evaluates an expression and decides what to do based on whether the result is:

- `True`
- `False`

The expression being evaluated is called a **condition**.

Operators themselves were covered in the previous notebook, so here we focus only on how their results are used for decision-making.

```python
candidate_score = 82

print(candidate_score >= 75)
print(candidate_score < 75)
```

# 2. The `if` Statement

The simplest conditional structure is an `if` statement.

The indented block executes only when the condition evaluates to `True`.

### Syntax

    if condition:
        statements

Python uses **indentation** to define the block belonging to the condition.

```python
candidate_score = 82

if candidate_score >= 75:
    print("Candidate cleared the assessment.")

print("Assessment processing completed.")
```

# 3. `if-else`

An `else` block provides an alternative action when the `if` condition evaluates to `False`.

This is useful when exactly two possible decision paths exist.

```python
experience_years = 4

if experience_years >= 3:
    print("Candidate meets the experience requirement.")
else:
    print("Candidate does not meet the experience requirement.")
```

# 4. `if-elif-else`

Real-world decisions often involve more than two possible outcomes.

`elif` means **else if** and allows additional conditions to be evaluated sequentially.

Python checks conditions from top to bottom and executes the **first matching branch**.

```python
score = 84

if score >= 90:
    category = "Excellent"
elif score >= 75:
    category = "Good"
elif score >= 60:
    category = "Average"
else:
    category = "Needs Improvement"

print(category)
```

## Why Condition Order Matters

Only the first matching branch in an `if-elif-else` chain executes.

Therefore, more restrictive or higher-priority conditions should normally appear before broader conditions.

For example, if `score >= 60` appears before `score >= 90`, a score of 95 would match the first condition and never reach the 90+ category.

```python
score = 95

# Incorrect ordering
if score >= 60:
    result = "Passed"
elif score >= 90:
    result = "Outstanding"

print(result)
```

# 5. Multiple Conditions

Business rules frequently depend on several requirements at the same time.

Logical operators such as `and`, `or`, and `not` can combine conditions.

The operators themselves were covered previously; here we use them specifically to construct decision rules.

```python
assessment_score = 82
experience_years = 4
background_check = True

if assessment_score >= 75 and experience_years >= 3 and background_check:
    print("Candidate is eligible.")
else:
    print("Candidate does not meet all requirements.")
```

# 6. Membership Checks in Conditions

Membership tests are useful when a decision depends on whether a value belongs to an approved or restricted collection.

Typical examples include:

- approved employee locations,
- supported API status values,
- permitted user roles,
- valid model labels.

```python
employee_location = "Hyderabad"
approved_locations = ["Hyderabad", "Bengaluru", "Pune"]

if employee_location in approved_locations:
    print("Employee belongs to an approved location.")
else:
    print("Location requires additional review.")
```

# 7. Chained Comparisons

Python allows comparisons to be written naturally using chained comparison syntax.

Instead of:

`score >= 60 and score <= 100`

Python allows:

`60 <= score <= 100`

This usually improves readability.

```python
score = 78

if 60 <= score <= 100:
    print("Score falls within the valid passing range.")
```

# 8. Truthy and Falsy Values in Conditions

Python allows values themselves to be evaluated as conditions.

Common falsy values include:

- `False`
- `None`
- `0`
- empty strings
- empty collections

This behavior was introduced with Python's type system earlier. Here we use it only for concise decision-making.

```python
candidate_name = "Neha"

if candidate_name:
    print("Candidate name is available.")
else:
    print("Candidate name is missing.")
```

```python
api_data = []

if api_data:
    print("API returned records.")
else:
    print("No records returned.")
```

# 9. Nested Conditional Statements

An `if` statement can exist inside another conditional block.

Nested conditions are useful when a second decision only makes sense after the first condition succeeds.

However, excessive nesting can make code difficult to read and maintain.

```python
assessment_score = 88
background_check = True

if assessment_score >= 75:
    if background_check:
        print("Candidate is ready for the next stage.")
    else:
        print("Assessment passed, but background verification is pending.")
else:
    print("Assessment score did not meet the requirement.")
```

## Prefer Simpler Conditions When Possible

Sometimes nested conditions can be replaced by a combined condition.

Instead of deeply nesting several `if` blocks, ask whether the business rule can be expressed clearly using a single condition.

Readable code is usually more important than minimizing the number of lines.

```python
assessment_score = 88
background_check = True

if assessment_score >= 75 and background_check:
    print("Candidate is ready for the next stage.")
else:
    print("Candidate requires further review.")
```

# 10. Conditional Expressions — Ternary Operator

Python provides a compact way to select between two values.

### Syntax

`value_if_true if condition else value_if_false`

Conditional expressions are best used for **simple decisions**.

Complex decision logic should remain in normal `if-else` blocks for readability.

```python
model_confidence = 0.89

prediction_status = (
    "High confidence"
    if model_confidence >= 0.80
    else "Manual review required"
)

print(prediction_status)
```

# 11. Guard-Style Logic

Guard-style logic checks exceptional or invalid situations early.

Instead of wrapping the entire main logic inside multiple nested conditions, problematic cases can be handled first.

This style becomes especially useful inside functions, APIs, validation pipelines, and production applications.

```python
candidate_score = 105

if not 0 <= candidate_score <= 100:
    print("Invalid score.")
else:
    print("Score is valid and can be processed.")
```

# 12. Practical Example — Candidate Eligibility

Suppose a candidate must satisfy the following requirements:

- Assessment score of at least 75
- At least 3 years of experience
- Required skill must be available
- Background verification must be complete

This combines multiple business conditions into a single decision.

```python
assessment_score = 86
experience_years = 5
candidate_skills = ["Python", "SQL", "Machine Learning"]
background_verified = True

required_skill = "Python"

if (
    assessment_score >= 75
    and experience_years >= 3
    and required_skill in candidate_skills
    and background_verified
):
    decision = "Eligible"
else:
    decision = "Not Eligible"

print(f"Candidate Decision: {decision}")
```

# 13. Practical Example — Employee Classification

Conditional statements can translate business rules into categories.

Here an employee's performance score determines the performance classification.

```python
performance_score = 4.3

if performance_score >= 4.5:
    classification = "Exceptional"
elif performance_score >= 3.5:
    classification = "High Performer"
elif performance_score >= 2.5:
    classification = "Meets Expectations"
else:
    classification = "Development Required"

print(f"Performance Classification: {classification}")
```

# 14. Practical Example — API Response Handling

Applications interacting with APIs frequently make decisions based on HTTP response codes.

Conditional logic can determine whether processing should continue, retry, request authentication, or report an error.

```python
status_code = 404

if 200 <= status_code < 300:
    message = "Request successful"
elif status_code == 401:
    message = "Authentication required"
elif status_code == 404:
    message = "Resource not found"
elif 500 <= status_code < 600:
    message = "Server-side error"
else:
    message = "Unexpected response"

print(message)
```

# 15. Practical Example — Model Confidence Threshold

Machine-learning systems often attach confidence or probability scores to predictions.

Conditional logic can translate those scores into operational actions.

For example:

- high confidence → automate,
- moderate confidence → human review,
- low confidence → reject or request additional evidence.

```python
confidence = 0.76

if confidence >= 0.90:
    action = "Automatically accept prediction"
elif confidence >= 0.70:
    action = "Send prediction for human review"
else:
    action = "Do not use prediction automatically"

print(action)
```

# 16. Common Conditional Mistakes

Some frequent issues include:

### 1. Incorrect indentation

Python uses indentation to determine which statements belong to a conditional block.

### 2. Incorrect condition ordering

Broad conditions can prevent more specific `elif` branches from executing.

### 3. Using unnecessary comparisons

Instead of:

`if is_active == True:`

prefer:

`if is_active:`

### 4. Excessive nesting

Several layers of nested `if` statements make business rules harder to understand.

### 5. Overusing ternary expressions

Ternary expressions are useful for simple decisions, but complex logic should use standard `if-elif-else`.

### 6. Forgetting boundary conditions

For thresholds, carefully consider whether rules require `>`, `<`, `>=`, or `<=`.

```python
is_active = True

# Less Pythonic
if is_active == True:
    print("Employee is active.")

# Preferred
if is_active:
    print("Employee is active.")
```

# 17. Readable Conditional Design

Readable decision logic should communicate business intent clearly.

Prefer:

- meaningful variable names,
- simple conditions,
- logical grouping,
- minimal nesting,
- clear threshold ordering.

Parentheses can also improve readability when several conditions are combined.

```python
score = 85
experience = 4
security_clearance = True

meets_assessment_requirement = score >= 75
meets_experience_requirement = experience >= 3

if (
    meets_assessment_requirement
    and meets_experience_requirement
    and security_clearance
):
    print("Candidate satisfies all requirements.")
```
