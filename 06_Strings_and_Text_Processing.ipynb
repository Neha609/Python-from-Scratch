## 1. What Is a String?

A **string** is a sequence of Unicode characters used to represent textual information.

Python strings are created using quotes:

- Single quotes: `'Python'`
- Double quotes: `"Python"`
- Triple quotes: `'''Python'''` or `"""Python"""`

Single and double quotes behave the same. The choice generally depends on readability and the text being represented.

Triple quotes are especially useful when text spans multiple lines.

```python
skill = "Python"
role = 'Data Scientist'

prompt = """Analyze the candidate profile
and identify the most relevant skills."""

print(skill)
print(role)
print(prompt)

print(type(skill))
```

## 2. String Indexing

Strings are ordered sequences of characters.

Each character has a position called an **index**.

Python uses **zero-based indexing**, meaning the first character is at position `0`.

Python also supports negative indexing:

- `-1` → last character
- `-2` → second-last character
- and so on

This is useful when extracting specific characters from labels, identifiers, filenames, or structured text.

```python
text = "Generative AI"

print("First character:", text[0])
print("Second character:", text[1])

print("Last character:", text[-1])
print("Second-last character:", text[-2])
```

## 3. String Slicing

Slicing allows us to extract a portion of a string.

The general syntax is:

`string[start:stop:step]`

Important points:

- `start` is included
- `stop` is excluded
- `step` controls how characters are selected
- Omitting `start` means start from the beginning
- Omitting `stop` means continue until the end

Slicing is especially useful when dealing with identifiers, prefixes, suffixes, and structured text.

```python
text = "Machine Learning"

print(text[0:7])
print(text[:7])
print(text[8:])
print(text[::2])
print(text[::-1])
```

### Practical Example — Extracting Structured Text

Suppose an application generates request IDs such as:

`REQ-2026-10458`

We may want to extract different portions of the identifier without performing more complex processing.

```python
request_id = "REQ-2026-10458"

request_type = request_id[:3]
request_year = request_id[4:8]
request_number = request_id[-5:]

print("Request Type:", request_type)
print("Year:", request_year)
print("Request Number:", request_number)
```

## 4. Strings Are Immutable

Python strings are **immutable**.

This means that once a string is created, its individual characters cannot be modified directly.

For example, we cannot replace a character using an index.

Instead, Python creates a **new string** whenever we perform a transformation.

This behavior is important because most string methods do not modify the original string.

```python
skill = "python"

# This would raise an error:
# skill[0] = "P"

corrected_skill = "P" + skill[1:]

print("Original:", skill)
print("New:", corrected_skill)
```

## 6. Commonly Used String Methods

Python strings provide many built-in methods for text manipulation.

Some of the most useful methods include:

| Method | Purpose |
|---|---|
| `lower()` | Convert text to lowercase |
| `upper()` | Convert text to uppercase |
| `title()` | Convert words to title case |
| `capitalize()` | Capitalize the beginning |
| `strip()` | Remove surrounding whitespace |
| `replace()` | Replace text |
| `split()` | Break text into parts |
| `join()` | Combine text values |
| `find()` | Find text position |
| `startswith()` | Check beginning of string |
| `endswith()` | Check end of string |

Rather than memorizing every available string method, focus on understanding the methods that repeatedly appear in real-world text-processing workflows.

```python
raw_skill = "   MACHINE LEARNING   "

print("Original:", repr(raw_skill))

print("Lower:", raw_skill.lower())
print("Upper:", raw_skill.upper())
print("Title:", raw_skill.title())
print("Cleaned:", raw_skill.strip())
print("Cleaned + Title:", raw_skill.strip().title())
```

## 7. Searching Inside Strings

When processing text, we frequently need to determine whether a word or phrase exists.

Python provides multiple ways to search strings.

### `in`

Use `in` when you only need to know whether a substring exists.

### `find()`

Returns the index of the first occurrence.

If the substring is not found, it returns `-1`.

### `index()`

Similar to `find()`, but raises an error if the substring does not exist.

For many text-processing workflows, `in` and `find()` are convenient because they allow us to handle missing values safely.

```python
document_text = "Candidate has strong Python, SQL, and machine learning experience."

print("Python" in document_text)

print("SQL position:", document_text.find("SQL"))
print("Java position:", document_text.find("Java"))
```

### `startswith()` and `endswith()`

Sometimes we do not want to search the complete string.

Instead, we want to check whether text starts or ends with a particular pattern.

Typical applications include:

- Checking file extensions
- Validating prefixes
- Identifying API endpoints
- Checking labels
- Validating document names

```python
filename = "candidate_resume.pdf"
api_endpoint = "/api/candidates"

print(filename.endswith(".pdf"))
print(filename.endswith(".csv"))

print(api_endpoint.startswith("/api"))
```

## 8. Replacing Text

The `replace()` method replaces one piece of text with another.

This is useful when:

- Standardizing labels
- Cleaning known patterns
- Formatting API values
- Preparing text for display
- Creating filenames
- Cleaning user-entered values

Because strings are immutable, `replace()` returns a new string rather than changing the original one.

```python
skills = "Python | SQL | Machine Learning"

cleaned_skills = skills.replace(" | ", ", ")

print(cleaned_skills)

api_status = "APPROVAL_PENDING"

display_status = api_status.replace("_", " ").title()

print(display_status)
```

## 9. Splitting Strings

The `split()` method breaks one string into multiple text components.

For example:

`"Python,SQL,GenAI"`

can be separated into individual skill values.

Splitting is extremely common when processing:

- Candidate skills
- Tags
- Comma-separated values
- API strings
- User input
- Labels and categories

By default, `split()` separates text using whitespace, but a specific delimiter can also be provided.

```python
skills_text = "Python,SQL,Machine Learning,GenAI"

skills = skills_text.split(",")

print(skills)

sentence = "Python makes text processing simple"

words = sentence.split()

print(words)
```

## 10. Joining Strings

`join()` performs the opposite type of operation.

Instead of breaking one string into multiple pieces, it combines multiple strings into one.

The syntax follows:

`separator.join(values)`

Common separators include:

- `", "` for comma-separated text
- `" "` for sentences
- `"\n"` for multiline output
- `" | "` for formatted labels

`join()` is especially useful when constructing prompts, context blocks, output messages, and summaries.

```python
skills = ["Python", "SQL", "Machine Learning", "GenAI"]

comma_separated = ", ".join(skills)

print(comma_separated)

prompt_context = "\n".join(skills)

print("\nPrompt Context:")
print(prompt_context)
```

## 11. Removing Unwanted Whitespace

Whitespace problems are extremely common in real-world data.

Text may contain:

- Leading spaces
- Trailing spaces
- Extra line breaks
- Tabs

Python provides three important methods:

- `strip()` → removes whitespace from both sides
- `lstrip()` → removes whitespace from the left
- `rstrip()` → removes whitespace from the right

Cleaning surrounding whitespace should often be one of the first steps when processing user input or external text.

```python
user_input = "   machine learning engineer   "

print("Original:", repr(user_input))

print("strip():", repr(user_input.strip()))
print("lstrip():", repr(user_input.lstrip()))
print("rstrip():", repr(user_input.rstrip()))
```

## 12. Case Transformation

The same textual value may appear in multiple forms:

- `PYTHON`
- `Python`
- `python`

If capitalization is not important to the business meaning, standardizing case makes comparison and processing easier.

Useful methods include:

- `lower()`
- `upper()`
- `title()`
- `capitalize()`
- `casefold()`

`casefold()` is particularly useful for stronger Unicode-aware case-insensitive comparisons.

```python
skill_1 = "PYTHON"
skill_2 = "python"

print(skill_1.lower() == skill_2.lower())

job_title = "senior machine learning engineer"

print(job_title.upper())
print(job_title.title())
print(job_title.capitalize())
```

### `lower()` vs `casefold()`

For normal English text, `lower()` is usually sufficient.

However, applications working with multilingual text may require more robust case-insensitive comparison.

`casefold()` performs stronger Unicode-aware normalization than `lower()`.

This distinction becomes more important in international applications and multilingual text processing.

```python
text_1 = "Straße"
text_2 = "STRASSE"

print(text_1.lower() == text_2.lower())
print(text_1.casefold() == text_2.casefold())
```

## 14. Formatting Strings with f-Strings

f-strings are the modern and preferred way to insert values into strings.

An f-string starts with the prefix `f`.

Variables or expressions can then be placed inside `{}`.

They are widely used when building:

- Messages
- Reports
- Prompts
- Logs
- Labels
- API outputs
- Model summaries

f-strings are generally more readable than manual string concatenation.

```python
candidate = "Neha"
experience = 10
primary_skill = "Machine Learning"

summary = (
    f"{candidate} has {experience} years of experience "
    f"with expertise in {primary_skill}."
)

print(summary)
```

## 15. Formatting Numbers and Text

f-strings also allow us to control how values are displayed.

Common formatting requirements include:

- Decimal precision
- Percentages
- Thousand separators
- Alignment

This is useful when generating reports, model metrics, dashboards, logs, or user-facing messages.

```python
model_accuracy = 0.94357
records_processed = 7250000
candidate_score = 91.456

print(f"Accuracy: {model_accuracy:.2%}")
print(f"Records Processed: {records_processed:,}")
print(f"Candidate Score: {candidate_score:.1f}")
```

## 19. Unicode Basics

Python 3 strings use **Unicode**.

Unicode allows Python strings to represent characters from many languages as well as symbols and emojis.

This is important because modern applications frequently process:

- International names
- Multilingual documents
- Currency symbols
- User-generated content
- AI-generated multilingual text

Therefore, a Python string is better understood as a sequence of Unicode characters rather than simply a sequence of English letters.

```python
english = "Hello"
hindi = "नमस्ते"
currency = "₹"
emoji = "🤖"

print(english)
print(hindi)
print(currency)
print(emoji)
```

## 20. Text Normalization Basics

Text that appears visually similar may sometimes be represented differently internally.

Unicode normalization converts text into a more consistent representation.

Python provides the built-in `unicodedata` module for this purpose.

A commonly useful normalization form is `NFKC`.

This can help when processing text collected from different systems, documents, browsers, or user inputs.

```python
import unicodedata

text = "ＡＩ Engineer"

normalized_text = unicodedata.normalize("NFKC", text)

print("Original:", text)
print("Normalized:", normalized_text)
```

## 21. Practical Text-Cleaning Pattern

In real applications, text cleaning rarely depends on only one method.

Multiple simple transformations are often combined.

Suppose candidate skills arrive as:

`"  PYTHON, sql , Machine Learning , GENAI  "`

A sensible cleaning workflow could be:

1. Remove unnecessary outer whitespace
2. Split the text into individual skills
3. Remove spaces around every skill
4. Standardize capitalization
5. Combine the cleaned values again

This demonstrates how simple string operations can form an effective text-processing pipeline.

```python
raw_skills = "  PYTHON, sql , Machine Learning , GENAI  "

skills = raw_skills.strip().split(",")

cleaned_skills = [
    skill.strip().title()
    for skill in skills
]

final_skills = ", ".join(cleaned_skills)

print("Original:", raw_skills)
print("Cleaned:", final_skills)
```

## 22. Practical Example — Cleaning User Input

User-entered text commonly contains inconsistent capitalization and unnecessary spaces.

Before comparing or storing such values, basic normalization can make the input more consistent.

For example:

`"   MACHINE LEARNING Engineer  "`

could be converted into a standardized representation before further processing.

```python
user_input = "   MACHINE LEARNING Engineer  "

normalized_input = user_input.strip().casefold()

print("Original:", repr(user_input))
print("Normalized:", normalized_input)
```

## 27. String Performance Considerations

Strings are immutable.

Therefore, string transformations generally create new string objects.

For small amounts of text, this is rarely a concern.

However, repeatedly concatenating a very large number of strings using `+` may create unnecessary intermediate strings.

When many pieces of text need to be combined, `join()` is usually the preferred approach.

This consideration becomes more relevant when processing:

- Large documents
- Thousands of records
- Large prompt contexts
- Model outputs
- Text-processing pipelines

Performance optimization should still be driven by actual application requirements rather than premature optimization.

```python
prompt_parts = [
    "Candidate Profile",
    "Python Developer",
    "8 Years Experience",
    "Machine Learning",
    "Generative AI"
]

prompt_context = "\n".join(prompt_parts)

print(prompt_context)
```

## 28. Common Mistakes

### Mistake 1 — Trying to modify a string directly

Strings are immutable, so individual characters cannot be changed using indexing.

### Mistake 2 — Forgetting that methods usually return new strings

Calling `text.strip()` does not permanently change `text` unless the returned value is stored.

### Mistake 3 — Forgetting whitespace during comparison

`"Python"` and `" Python "` are different strings.

### Mistake 4 — Assuming `find()` returns `None`

When text is not found, `find()` returns `-1`.

### Mistake 5 — Using unnecessary manual logic

Many text-processing problems are already solved cleanly by built-in methods such as `split()`, `join()`, `replace()`, `strip()`, and `startswith()`.

### Mistake 6 — Excessive method chaining

Method chaining is useful, but extremely long chains can reduce readability.

### Mistake 7 — Assuming every text comparison should use `lower()`

For multilingual case-insensitive matching, `casefold()` may sometimes be more appropriate.
