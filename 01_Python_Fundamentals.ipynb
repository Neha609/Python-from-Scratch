## 📌 1. What is Python?

A simple Python program looks like:


Python hides much of the low-level complexity involved in:

* Memory allocation
* Machine instructions
* Explicit type declarations

## 🔹 5. Indentation and Code Blocks

Python uses **indentation** to define blocks of code.



## 🔹 6. Statements vs Expressions

Understanding this distinction helps explain how Python evaluates code.

### Expression

An **expression produces a value**.

```python

price * quantity
score >= 80
```

### Statement

A **statement performs an instruction**.

```python
total = 100

    print(total)
```

### Mental Model

```text
Expression → Produces a value

Statement  → Performs an instruction
```

Example:

```python

```

Here:

* `10 + 20` → expression
* `total = 10 + 20` → assignment statement

> 💡 **Good to Know:** Expressions often appear inside statements.


````python
# Expression
10 + 20
````

````python
# Assignment statement containing an expression
total = 10 + 20

print(total)
````

## 🔍 10. Dynamic Typing

Python does not permanently attach one type to a variable name.

Consider:

```python
value = 100
value = "Python"
```

Conceptually:

```text
Initially:

value ─────► 100
             int object


Later:

value ─────► "Python"
             str object
```

The **objects have types**.

The name `value` can be rebound to another object.

> 🎯 **Interview Point:** A useful mental model is that Python names refer to objects, and the objects themselves have types.


````python
value = 100

print(value)
print(type(value))

# The same name now refers to a string object
value = "Python"

print(value)
print(type(value))
````

## 🔹 11. Multiple Assignment, Swapping & Constants

Python provides several convenient assignment patterns.

### Multiple assignment

```python

```

### Assign the same value

```python
x = y = z = 0
```

### Swap values

```python
a, b = b, a
```

This avoids manually creating a temporary variable.

---

### Constants

Python does not strictly enforce constants.

Instead, uppercase names communicate that a value should normally remain unchanged.

```python
MAX_RETRIES = 3
API_TIMEOUT_SECONDS = 30
PI = 3.14159
```

> 💡 **Good to Know:** Uppercase naming is a convention, not runtime enforcement.


````python
name, experience, domain = "Neha", 5, "AI"

print(name)
print(experience)
print(domain)

# Pythonic swapping
first = 10
second = 20

first, second = second, first

print(first, second)

# Constant by convention
MAX_RETRIES = 3
print(MAX_RETRIES)
````

# 🔍 How Python Works Internally

## Names, Objects & References

One of the most important mental models in Python is:

```text
Name ─────► Object
```

Consider:

```python
score = 95
```

Conceptually:

```text
score ─────────► 95
                 │
                 └── int object
```

Python manages objects in memory, while names allow our code to refer to those objects.

### Multiple Names Can Reference the Same Object

```python
a = 1000
b = a
```

Conceptually:

```text
a ────────┐
           ▼
         1000
           ▲
b ────────┘
```

Both names currently refer to the same object.

> 🔍 **Under the Hood:** Assignment usually binds a name to an object; it does not automatically create an independent copy of that object.


````python
score = 95

print("Type:", type(score))
print("Identity:", id(score))

a = 1000
b = a

print("\na == b:", a == b)
print("a is b:", a is b)
print("id(a):", id(a))
print("id(b):", id(b))
````

## 🔍 `==` vs `is`

These operators answer different questions.

### Equality

```python
a == b
```

asks:

> Do these objects have equal values?

### Identity

```python
a is b
```

asks:

> Are both references pointing to the same object?

| Operator | Checks          |
| -------- | --------------- |
| `==`     | Value equality  |
| `is`     | Object identity |

### Important `None` Pattern

Prefer:


rather than:


⭐ **Must Know:** Do not use `is` for ordinary string or numeric value comparison.

Use:

```python
candidate_status == "Selected"
```

not:



````python
result = None

print(result is None)

candidate_status = "Selected"

print(candidate_status == "Selected")
````

## 🔍 Object Rebinding & Memory Management

Consider:

```python
message = "Python"
message = "AI"
```

Conceptually:

```text
Before:

message ─────► "Python"


After:

message ─────► "AI"

               "Python"
```

The name `message` has been **rebound**.

If an object becomes unreachable, Python's memory-management system can eventually reclaim its memory.

In CPython:

* Reference counting plays an important role.
* A cyclic garbage collector helps deal with certain reference cycles.

> 💡 **Good to Know:** Python developers normally do not manually allocate and free ordinary Python objects.

## 🎯 Floating-Point Precision

Consider:

```python
0.1 + 0.2
```

You might expect exactly:

```text
0.3
```

However, many decimal fractions cannot be represented exactly using binary floating-point.

Therefore:

```python
0.1 + 0.2 == 0.3
```

may evaluate to `False`.

This behavior is not unique to Python.

### Approximate Comparison

Python provides:

```python
math.isclose()
```

when approximate numeric comparison is appropriate.

For use cases requiring exact decimal arithmetic, such as some financial calculations, Python provides:

```python
decimal.Decimal
```

> ⚠️ **Common Mistake:** Do not assume floating-point calculations always produce exact decimal values.


````python
import math

result = 0.1 + 0.2

print("Result:", result)
print("Exact equality:", result == 0.3)
print("Approximate equality:", math.isclose(result, 0.3))
````

# ⚖️ Concept Comparisons

## `=` vs `==` vs `is`

| Operator | Meaning         | Example     |
| -------- | --------------- | ----------- |
| `=`      | Assignment      | `x = 10`    |
| `==`     | Value equality  | `x == 10`   |
| `is`     | Object identity | `x is None` |

---

## Dynamic Typing vs Strong Typing

| Concept        | Meaning in Python                                                                 |
| -------------- | --------------------------------------------------------------------------------- |
| Dynamic typing | Names do not require permanently declared types                                   |
| Strong typing  | Python generally avoids arbitrary implicit conversions between incompatible types |

Dynamic typing allows:

```python
value = 10
value = "Python"
```

Strong typing means this fails:

```python
"10" + 5
```

Python does not silently convert `5` into `"5"`.

---

## `/` vs `//`

| Operator | Purpose        | Example  | Result |
| -------- | -------------- | -------- | ------ |
| `/`      | True division  | `7 / 2`  | `3.5`  |
| `//`     | Floor division | `7 // 2` | `3`    |

Be careful with negative values:

```python
-7 // 2
```

returns:

```text
-4
```

because floor division rounds toward negative infinity.

---

## Comment vs Docstring

| Feature                     | Comment      | Docstring                                |
| --------------------------- | ------------ | ---------------------------------------- |
| Syntax                      | `# comment`  | String literal in documentation position |
| Runtime object              | No           | Yes                                      |
| Purpose                     | Explain code | Document modules, functions, classes     |
| Programmatically accessible | No           | Yes                                      |


````python
print("7 / 2   =", 7 / 2)
print("7 // 2  =", 7 // 2)

print("-7 / 2  =", -7 / 2)
print("-7 // 2 =", -7 // 2)
````

# 🎯 Interview Preparation

### 1. Is Python compiled or interpreted?

**Answer:** Python is commonly described as interpreted, but CPython first compiles source code into bytecode, which is then executed by the Python runtime.

---

### 2. What is dynamic typing?

**Answer:** Python names do not require fixed type declarations. The same name can refer to objects of different types during execution.

---

### 3. Is Python strongly typed?

**Answer:** Yes. Python generally does not silently perform arbitrary conversions between incompatible types.

For example:

```python
"10" + 5
```

raises `TypeError`.

---

### 4. What is the difference between `=` and `==`?

`=` performs assignment.

`==` checks value equality.

---

### 5. What is the difference between `==` and `is`?

`==` compares values.

`is` compares object identity.

---

### 6. What does `input()` return?

`input()` always returns a `str`.

---

### 7. Why does Python use indentation?

Indentation defines code blocks as part of Python's syntax and encourages consistent structure.

---

### 8. What is CPython?

CPython is the reference and most widely used implementation of Python, implemented primarily in C.

---

### 9. What is Python bytecode?

Bytecode is an intermediate instruction representation produced by implementations such as CPython and executed by the Python runtime.

---

### 10. Why should `None` normally be checked using `is None`?

`None` is a singleton object, so identity checking is the conventional and intended approach.

---

### 11. Why should names such as `list`, `str`, and `sum` be avoided for variables?

They shadow Python built-ins and can make later code confusing or incorrect.

---

### 12. Why can `0.1 + 0.2 == 0.3` be false?

Because many decimal fractions cannot be represented exactly using binary floating-point.

---

### 13. What is truthiness?

Truthiness is Python's ability to evaluate objects in Boolean contexts without explicitly comparing them to `True` or `False`.

---

### 14. Why does notebook execution order matter?

Jupyter/Colab kernels maintain state between executed cells. A cell may therefore depend on variables created by an earlier executed cell.

> 🎯 **Interview Point:** Focus on explaining **why Python behaves this way**, rather than memorizing one-line definitions.
